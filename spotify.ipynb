# What is The Weeknd's Average Number of Streams in 2023 on Spotify?

To find this out, I will be conducting a **census study** because I will be counting the total number of streams with songs by only The Weeknd and then finding the **mean** of his total streams. I will be analyzing this using **quantitative data** so that I can isolate the songs by just The Weeknd and then look at the number of streams on only the popular songs released in 2023. 

```python
import pandas as pd

spotify = pd.read_csv("spotify-2023.csv",encoding="latin-1")
spotify.head()
```

## Overview:

```python
spotify.info()
```

## Songs by The Weeknd:

```python
spotify[spotify["artist(s)_name"] == "The Weeknd"]
```

# Analysis:

## Average Number of Streams:

```python
weeknd = spotify[spotify["artist(s)_name"] == "The Weeknd"]
weeknd["streams"].astype(int).mean()
```

It was found to be **644797857.7272727**

## 5 Number Summary:

```python
streams = weeknd["streams"].astype(int)
print("Minimum:", streams.min())
print("Q1:", streams.quantile(0.25))
print("Median:", streams.median())
print("Q3:", streams.quantile(0.75))
print("Maximum:", streams.max())
```

# Visual:

```python
import matplotlib.pyplot as plt

plt.bar(weeknd["track_name"], weeknd["streams"].astype(int))
plt.xlabel("Song")
plt.ylabel("Number of Streams")
plt.title("The Weeknd's Streams in 2023")
plt.xticks(rotation=90)
plt.show()
```

# Conclusion:

After conducting this evaluation of finding the **mean** number of streams by The Weeknd on Spotify in 2023, I have found the average to be around **644,797,858 streams.** Looking at the graph, the distribution isn't very symmetrical because the song "Blinding Lights" surpasses all of his other songs drastically. I'd say that that some songs have much higher bars on the graph due to those specific songs being more popular, therefore, getting more streams. Since there are songs that are getting way more streams than others, his average number of streams is increased by a lot more than expected.
